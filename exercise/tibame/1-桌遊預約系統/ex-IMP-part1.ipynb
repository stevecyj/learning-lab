```python
games = {"小瓦隆": None, "雨聲": None, "龍捲風": None}

current_user_id = "123abc"


def find_game_names(message):
    """找出出現在使用者訊息中的遊戲名稱。"""
    matched_games = [game_name for game_name in games if game_name in message]
    return matched_games


def end_service():
    """服務結束"""
    print("服務結束")


def cancel_reservations():
    """取消目前使用者的所有預約。"""
    canceled_games = []

    for game_name, user_id in games.items():
        if user_id == current_user_id:
            games[game_name] = None
            canceled_games.append(game_name)

    if len(canceled_games) == 0:
        print("你目前沒有預約。")
    else:
        game_text = "、".join(canceled_games)
        print(f"已取消預約：{game_text}")


def show_available_games():
    """顯示所有尚未被預約的遊戲。"""
    available_games = [
        game_name for game_name, user_id in games.items() if user_id is None
    ]

    if len(available_games) == 0:
        print("目前沒有可預約的遊戲。")
    else:
        game_text = "\n".join(available_games)
        print(f"目前可預約的遊戲：\n{game_text}")


def show_unknown_command():
    """無法理解指令"""
    print("我不明白這個指令。")
    print("你可以查看遊戲、預約遊戲、取消預約，或輸入「結束」。")


def reserve_game(message):
    """從訊息中找出遊戲，然後為目前使用者預約。"""
    matched_games = find_game_names(message)

    if len(matched_games) == 0:
        print("找不到遊戲名稱，請說明想預約哪一款遊戲。")
        return

    if len(matched_games) > 1:
        game_text = "、".join(matched_games)
        print(f"你提到了多款遊戲：{game_text}，請一次指定一款。")
        return

    game_name = matched_games[0]
    reservation_user_id = games[game_name]

    if reservation_user_id is None:
        games[game_name] = current_user_id
        print(f"已為你預約：{game_name}")
    elif reservation_user_id == current_user_id:
        print(f"你已經預約過：{game_name}")
    else:
        print(f"{game_name} 已被其他使用者預約。")


while True:
    msg = input("輸入訊息: ").strip()

    if msg == "結束":
        end_service()
        break
    elif "取消" in msg:
        cancel_reservations()
    elif "預約" in msg or "訂" in msg:
        reserve_game(msg)
    elif "遊戲" in msg or "有哪些" in msg:
        show_available_games()
    else:
        show_unknown_command()
```
