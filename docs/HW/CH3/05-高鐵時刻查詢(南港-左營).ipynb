```python
import json
from urllib.parse import parse_qsl

import requests
from bs4 import BeautifulSoup

raw_payload = """
SearchType=S&Lang=TW&StartStation=NanGang&EndStation=ZuoYing&OutWardSearchDate=2026%2F08%2F29&OutWardSearchTime=10%3A00&ReturnSearchDate=2026%2F08%2F29&ReturnSearchTime=10%3A00&DiscountType=
""".strip()

payload = {}
for field_name, field_value in parse_qsl(raw_payload, keep_blank_values=True):
    payload[field_name] = field_value

url = "https://www.thsrc.com.tw/TimeTable/Search"
# 將前面組好的 payload 以 POST 方法送到高鐵時刻查詢 API。
response = requests.post(url, data=payload, verify=False)

# response.text 是伺服器回傳的 JSON 格式字串。
# json.loads 會把字串轉成 Python 字典，才能依欄位名稱取資料。
result = json.loads(response.text)

# data 是 API 的主要回傳內容；DepartureTable 的 TrainItem 是去程班次清單。
# 每次迴圈的 train 都代表一個班次，內容是一個字典。
for train in result["data"]["DepartureTable"]["TrainItem"]:
    # 用 f-string 從目前班次取出車次、出發時間與抵達時間，並輸出成一行。
    print(
        f"Train number: {train['TrainNumber']}, "
        f"Departure time: {train['DepartureTime']}, "
        f"Destination time: {train['DestinationTime']}"
    )
```
